```python
import pandas as pd

# 1. Load dataset
df = pd.read_csv("ott_movies_clean_unique.csv")


def format_duration(total_minutes):
    hours = int(total_minutes) // 60
    minutes = int(total_minutes) % 60
    if hours > 0:
        return f"{hours}h {minutes}m ({int(total_minutes)} mins)"
    return f"{minutes}m"


# 2. Numbered choices for easy user input
type_options = {
    "1": ("All", None),
    "2": ("Movie", "Movie"),
    "3": ("Series", "Series"),
}

genre_options = {
    "1": ("All", None),
    "2": ("Action", "Action"),
    "3": ("Comedy", "Comedy"),
    "4": ("Crime", "Crime"),
    "5": ("Drama", "Drama"),
    "6": ("Fantasy", "Fantasy"),
    "7": ("Sci-Fi", "Sci-Fi"),
    "8": ("Thriller", "Thriller"),
}

# Question 1: Content Type (loops until valid)
print("\nChoose Content Type:")
for key, (label, _) in type_options.items():
    print(f"  {key}. {label}")

while True:
    type_choice = input("Enter number (1-3): ").strip()
    if type_choice in type_options:
        break
    print("Invalid input! Please select a valid number between 1 and 3.")

# Question 2: Genre (loops until valid)
print("\nChoose Genre:")
for key, (label, _) in genre_options.items():
    print(f"  {key}. {label}")

while True:
    genre_choice = input("Enter number (1-8): ").strip()
    if genre_choice in genre_options:
        break
    print("Invalid input! Please select a valid number between 1 and 8.")

# Question 3: Minimum Rating (loops until valid)
while True:
    try:
        min_rating = float(input("\nEnter minimum rating (5.0 - 9.5): ").strip())
        if 5.0 <= min_rating <= 9.5:
            break
        print("Invalid input! Please enter a rating between 5.0 and 9.5.")
    except ValueError:
        print("Invalid input! Please enter a valid number (e.g., 7.5).")

# Question 4: Number of Results (loops until valid)
while True:
    try:
        top_n = int(input("\nHow many top results to show? (e.g., 5): ").strip())
        if top_n > 0:
            break
        print("Invalid input! Please enter a number greater than 0.")
    except ValueError:
        print("Invalid input! Please enter a valid whole number (e.g., 5).")

# Map numbered choices
_, selected_type = type_options[type_choice]
_, selected_genre = genre_options[genre_choice]

# 3. Filter dataset
filtered = df[df["Rating"] >= min_rating]

if selected_type:
    filtered = filtered[filtered["Type"] == selected_type]

if selected_genre:
    filtered = filtered[filtered["Genre"] == selected_genre]

# 4. Sort and display results with hours & minutes
results = filtered.sort_values(by="Trending_score", ascending=False).head(top_n)

if results.empty:
    print("\nNo matching titles found. Try a lower rating or choose 'All'.")
else:
    print(f"\n=== Top {len(results)} Recommendations ===")
    for i, (_, row) in enumerate(results.iterrows(), start=1):
        duration_text = format_duration(row["Duration"])
        print(f"\n{i}. {row['Title']} ({row['Type']}, {row['Release_year']})")
        print(f"   Genre: {row['Genre']} | Platform: {row['Platform']} | Rating: {row['Rating']}")
        print(f"   Duration: {duration_text}")
```
